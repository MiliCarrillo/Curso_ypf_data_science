```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# 1. Prueba de Numpy y Pandas (Creación de datos)
np.random.seed(42)
datos = {
    'Mes': np.arange(1, 13),
    'Usuarios_Activos': np.random.randint(100, 500, 12)
}
df = pd.DataFrame(datos)

# 2. Prueba de Scikit-Learn (Entrenamos un modelo predictivo muy simple)
X = df[['Mes']]
y = df['Usuarios_Activos']
modelo = LinearRegression()
modelo.fit(X, y)
df['Tendencia'] = modelo.predict(X)

# 3. Prueba de Matplotlib y Seaborn (Visualización)
plt.figure(figsize=(10, 5))
sns.set_theme(style="darkgrid")

# Dibujamos los datos reales y la línea de tendencia
sns.lineplot(data=df, x='Mes', y='Usuarios_Activos', marker='o', label='Usuarios Reales', color='blue')
sns.lineplot(data=df, x='Mes', y='Tendencia', color='red', linestyle='--', label='Tendencia (Sklearn)')

plt.title('Prueba de Entorno: Data Science YPF')
plt.xlabel('Mes del Año')
plt.ylabel('Cantidad de Usuarios')
plt.xticks(np.arange(1, 13))
plt.legend()
plt.show()

print("¡Éxito! pandas, numpy, matplotlib, seaborn y scikit-learn están funcionando a la perfección.")
```
